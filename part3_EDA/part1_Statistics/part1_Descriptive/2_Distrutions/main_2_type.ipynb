```python
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_normal = stats.norm.rvs(size=100000, loc=0, scale=1)
```

```python
sns.displot(data_normal, bins=100, kde=False)
```

```python
data_uniform = stats.uniform.rvs(size=100000, loc=0, scale=10)
```

```python
sns.displot(data_uniform, bins=100, kde=False)
```

```python
data_benouli = stats.bernoulli.rvs(size=1000, p=0.2) # 80% 0 , 20% 1
print(data_benouli)
```

```python
sns.distplot(data_benouli, bins=100, kde=False)
```

```python
data_binom = stats.binom.rvs(size=1000, n=100, p=0.2) # میاد 100 بار برنولی رو انجام میده و همه رو جمع میکنه
print(data_binom)
```

```python
sns.displot(data_binom, bins=50, kde=False)
```

```python
import numpy as np
```

```python
p = 1 / 6
data_multinomial = np.random.multinomial(n=6, pvals=[p, p, p, p, p, p], size=1000) # مثلا p ها اعداد تاس که 
print(data_multinomial)
#[0 1 2 1 2 0] الان احتمال 0 1 و احتمال 1 2 
```

```python
icu = 0.3
surgery = 0.2
el = 0.5
data_multinomial = np.random.multinomial(n=1000, pvals=[icu, surgery, el], size=365)
print(data_multinomial)
```

```python
data_poison = np.random.poisson(lam=7, size=1000) #    مثلا سرور از 100 روز 7 روز خراب معمولا
print(data_poison)
# الان تو 100 روز اول 3 بار خراب و تو 100 روز دوم 7 بار و
```

```python
sns.displot(data_poison, kde=False)
```

```python
data_exponential = np.random.exponential(scale=2, size=1000)# مثلا اتقافی افتاده و میانگین فاصله اون دو اتفاق 2 دقیقه است
print(data_exponential)
```

```python
sns.displot(data_exponential)
```
